```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample FAQ Data from a website
faq_database = [
    {
        "question": "What is your return policy?",
        "answer": "We offer a 30-day money-back guarantee on all unused items in their original packaging."
    },
    {
        "question": "Do you ship internationally?",
        "answer": "Yes, we ship to over 50 countries worldwide. Shipping fees and delivery times vary by destination."
    },
    {
        "question": "How can I track my order?",
        "answer": "Once your order ships, you will receive an email with a tracking link. You can also track it from your account dashboard."
    },
    {
        "question": "Can I change or cancel my order?",
        "answer": "Orders can be changed or canceled within 1 hour of placement by contacting our support team via live chat."
    }
]

# Extract questions for matching
questions = [item["question"] for item in faq_database]

# Initialize the TF-IDF Vectorizer and fit it on the FAQ questions
vectorizer = TfidfVectorizer().fit(questions)
question_vectors = vectorizer.transform(questions)

def ask_faq_agent(query: str, threshold: float = 0.25):
    # Transform the user query to the same vector space
    query_vector = vectorizer.transform([query])

    # Calculate cosine similarity between query and all questions
    similarities = cosine_similarity(query_vector, question_vectors).flatten()

    # Find the best match
    best_match_idx = np.argmax(similarities)
    best_score = similarities[best_match_idx]

    print(f"User: {query}")
    if best_score >= threshold:
        matched_faq = faq_database[best_match_idx]
        print(f"Matched Question: \"{matched_faq['question']}\" (Confidence: {best_score:.2f})")
        print(f"Agent: {matched_faq['answer']}\n")
    else:
        print(f"Agent: I'm sorry, I couldn't find a confident match for your question. Would you like to be connected to a human agent?\n")

# Example Queries to test the agent offline
print("--- FAQ Agent Testing (No API Key Required) ---\n")
ask_faq_agent("How do I track my package?")
ask_faq_agent("Can I get a refund if I return it within 2 weeks?")
ask_faq_agent("Do you sell custom-made shoes?")
```
